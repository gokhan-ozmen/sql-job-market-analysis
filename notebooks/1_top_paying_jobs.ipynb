```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
```

```python
file = pd.read_csv('1_top_paying_jobs.csv')
```

```python
file["job_label"] = file["company_name"] + " — " + file["job_title"]
```

```python
plt.figure(figsize=(10, 6))

plt.barh(file["job_label"], file["salary_year_avg"])

plt.xlabel("Average Yearly Salary ($)")
plt.title("Top 10 Highest-Paying Remote Data Analyst Jobs")

plt.gca().invert_yaxis()

plt.gca().xaxis.set_major_formatter(FuncFormatter(lambda x, pos: f"${x/1000:.0f}K"))
plt.xlim(0, file["salary_year_avg"].max() * 1.12)
plt.tight_layout()

plt.savefig("1_top_paying_jobs.png", dpi=300, bbox_inches="tight")

plt.show()
```
