```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
```

```python
file = pd.read_csv("5_optimal_skills.csv")
```

```python
file
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(file["demand_count"], file["avg_salary"])

for i, row in file.iterrows():
    plt.annotate(
        row["skills"],
        (row["demand_count"], row["avg_salary"]),
        xytext=(5, 5),
        textcoords="offset points")
    
plt.xlabel("Demand Count")
plt.ylabel("Average Yearly Salary ($)")
plt.title("Most Optimal Skills for Data Analysts")

plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda y, pos: f"${y/1000:.0f}K"))

plt.tight_layout()

plt.savefig("5_optimal_skills.png", dpi=300, bbox_inches="tight")
plt.show()
```
