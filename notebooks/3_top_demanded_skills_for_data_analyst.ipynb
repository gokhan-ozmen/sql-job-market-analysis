```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
```

```python
file = pd.read_csv('3_top_demanded_data_analyst_skills.csv')
```

```python
file 
```

```python
plt.figure(figsize=(10, 6))

bars = plt.barh(file["skills"], file["demand_count"])
plt.bar_label(bars, fmt="{:,.0f}", padding=3)

plt.xlabel("Demand Count")
plt.title("Top 5 Most In-Demand Skills for Data Analysts")

plt.gca().invert_yaxis()

plt.gca().xaxis.set_major_formatter(FuncFormatter(lambda x, pos: f"{x/1000:.0f}K"))

plt.xlim(0, file["demand_count"].max() * 1.12)

plt.tight_layout()

plt.savefig("3_top_demanded_skills.png", dpi=300, bbox_inches="tight")

plt.show()
```
